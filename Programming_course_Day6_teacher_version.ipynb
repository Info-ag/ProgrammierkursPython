````python
## Aufgabe 1: Strings, Pangrams
# Überarbeitet aus:
# Übungsblatt 2, Programmieren - Sommersemester 2025, Karlsruher Institut für Technologie (KIT)

# Schreibe eine Methode isPangram(sentence), die überprüft,
# ob ein Satz alle Buchstaben des englischen Alphabets (a–z) mindestens einmal enthält. Die
# Groß- und Kleinschreibung soll dabei ignoriert werden. Die Methode gibt True zurück, wenn
# es sich um ein Pangram handelt, andernfalls False.
abc = "abcdefghijklmnopqrstuvwxyz"

def isPangram(sentence):
    sentence = sentence.lower()
    has_all = True
    for char in abc:
        has_all = has_all and (char in sentence)
    return has_all

print(isPangram("This is not a pangram.")) # False
print(isPangram("Pack my box with five dozen liquor jugs.")) # True
````

````python
## Aufgabe 2: Listen und Integers, Gefilterte Liste
# Überarbeitet aus:
# Übungsblatt 2, Programmieren - Sommersemester 2025, Karlsruher Institut für Technologie (KIT)

# Schreibe eine Methode greaterThan(values, threshold),
# welche alle Werte in der gegebenen values-Liste in einer neuen int-Liste zurückgibt, die echt
# größer als der gegebene threshold-Wert sind.

def greaterThan(values, threshold):
    result = []
    for element in values:
        if element >= threshold:
            result.append(element)
    return result

print(greaterThan([-5, 0, 5], 0))
````

````python
## Aufgabe 3: Listen und Integer, Überschneidungen
# Überarbeitet aus:
# Übungsblatt 2, Programmieren - Sommersemester 2025, Karlsruher Institut für Technologie (KIT)

# Schreibe eine Methode areDisjoint(first, second),
# welche überprüft, ob die beiden gegebenen Listen first und second keine gemeinsamen
# Elemente enthalten. Gib True zurück, wenn es keine Überschneidung gibt, sonst False.

def areDisjoint(first, second):
    for element1 in first:
        if element1 in second:
            return False
    return True

print(areDisjoint([3, 1, 4], [2, 5, 6]))
print(areDisjoint([3, 1, 4], [2, 4, 6]))
````

````python
## Aufgabe 4: Dictionaries
# Schreibe eine Methode mergeDictionaries(first, second),
# welche ein neues Dictionary erstellt, das alle Key-Value-Paare
# aus first und second enthält.
# Falls ein key in beiden Dictionaries Werten vorkommt, gebe ein leeres Dictionary zurück.

def mergeDictionaries(first, second):
    result = {}
    for key, value in first.items():
        result[key] = value
    for key, value in second.items():
        if key in result.keys(): 
            return {}
        result[key] = value
    return result

print(mergeDictionaries({"Deutschland": 84, "China": 1419}, {"Mali": 24, "Ukraine": 38}))
print(mergeDictionaries({"Deutschland": 84, "China": 1419}, {"Deutschland": 42, "Mali": 24, "Ukraine": 38})) # {}
````

Enums (kurz für enumerations, also Aufzählungen) sind eine Menge von Namen,


````python
class Player:
    def __init__(user_name, display_name, current_level, exp_count):
        self.user_name = user_name
        self.display_name = display_name
        self.current_level = current_level
        self.exp_count = exp_count
        self.inventory = [] # einfach eine Liste mit Gegenständen, wird gefüllt wenn Spieler Gegenstände sammelt
        self.hp = 100 # Für Schaden-Idee
        
    def pickup_object(new_object):
        self.inventory.append(new_object)
        
    def get_damage(damage): # Schaden-Idee
        number_of_armor_in_inventory = 0
        for obj in self.inventory:
            if obj.object_type == ObjectType.ARMOR:
                number_of_armor_in_inventory += 1
        damage /= number_of_armor_in_inventory + 1 # Unverändert wenn keine Rüstung
        
        self.hp -= damage
        
        if self.hp <= 0:
            print(f"Spieler {self.display_name} ist gestorben!")
            self.exp_count = 0
            self.inventory = []
````

````python
class Enemy:
    def __init__(self, name, hp, atk):
        self.name = name
        self.hp = hp
        self.atk = atk
        
class Goblin(Enemy):
    def __init__(self, name):
        super().__init__(name, 100, 15)

class Troll(Enemy):
    def __init__(self, name):
        super().__init__(name, 2000, 50)
        
class Drake(Enemy):
    def __init__(self, name):
        super().__init__(name, 5000, 250)
        
class Undead(Enemy):
    def __init__(self, name):
        super().__init__(name, 150, 30)
        
class Bandit(Enemy):
    def __init__(self, name):
        super().__init__(name, 50, 5)
````

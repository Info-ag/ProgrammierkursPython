## X. Klassen und Objekte
### Objekte

Heute geht es um Klassen und Objekte in Python.
Ein **Objekt** hat **Eigenschaften** (Dinge, die es besitzt) und **Fähigkeiten** (Dinge, die es kann).
- Die Eigenschaften werden auch **Attribute** genannt und sie werden meistens mit Substantiven bezeichnet. 
- Die Fähigkeiten eines Objekts sind seine **Methoden** (einfach Funktionen, die als Methoden bezeichnet werden). Sie werden meistens mit Verben benannt.

Anhand eines Fahrzeugs kann man sich das so vorstellen:

```python
from IPython import display
display.Image("Klassen_Übersicht.png")
```

Objekte können Methoden von anderen Objekten aufrufen, indem sie von Ihnen "erben". Ein Objekt **Auto**, das von Fahrzeug erbt, könnte also auch die Methoden beschleunige und bremsen verwenden (aber dazu gleich mehr).

### Klassen
Eine Klasse ist eine Art Bauplan mit dem Objekte erstellt werden können. Im Programmcode definiert man eine Klasse und dann lässt man Objekte dieser Klasse zusammenbauen. Das Zusammenbauen der Objekte nennt man auch **"Instanziieren"**. Je nachdem, welche Variablen man beim Instanziieren mitgibt, können verschiedene Objekte erstellt werden.

Wichtig ist dabei, dass zwei Objekte immer die gleichen Eigenschaften (Attribute) und die gleichen Fähigkeiten (Methoden) haben. Beispielsweise haben sowohl ein Auto als auch ein Zug beide eine Anzahl an Rädern, eine Farbe, sowie einen Kraftstoff als Attribut. Ebenso haben sie beide die gleichen Methoden (z.B. beschleunigen, bremsen, ...). **Die Methoden sind bei allen Objekten immer identisch.**
Bei den Attributen sieht das ein wenig anders aus. Zwar haben zwei Objekte immer die gleichen Attribute, allerdings können die **Werte jener Attribute sich zwischen den Objekten unterscheiden**. Beispielsweise haben sowohl ein Auto und ein Zug Räder, jedoch unterscheiden sie sich in der Anzahl an Rädern. In einem Objekt sind also die Daten (Attribute) und Fähigkeiten (Methoden) zusammengefasst.

### Definition einer Klasse 
Eine Klasse könnte nur wie folgt definiert werden:

```python
class Fahrzeug:
    """In dieser Zeichenkette steht eine Beschreibung des Objekts:
    Ein Fahrzeug hat eine Anzahl an Rädern, eine Farbe und
    eine Antriebstechnologie.
    Außerdem kann es beschleunigen und bremsen."""
    def __init__(self, anzahl_räder=0, farbe="schwarz"):
        self.anzahl_räder = anzahl_räder
        self.farbe = farbe

    def beschleunigen(self):
        ausgabe = "Ich bin ein Fahrzeug mit Anzahl an Rädern: "
        ausgabe += str(self.anzahl_räder)
        ausgabe += " und Farbe: "
        ausgabe += self.farbe
        ausgabe += " und ich beschleunige jetzt."
        print(ausgabe)

    def bremsen(self):
        ausgabe = "Ich bin ein Fahrzeug mit Anzahl an Rädern: "
        ausgabe += str(self.anzahl_räder)
        ausgabe += " und Farbe: "
        ausgabe += self.farbe
        ausgabe += " und ich bremse jetzt."
        print(ausgabe)
```

Unsere Klasse hat den Namen **Fahrzeug** und beinhaltet die 3 Methoden `__init__`, `beschleunigen` und `bremsen`.

Die Funktion `__init__` wird bei der Instanziierung des Objektes aufgerufen, sie "initialisiert" also das Objekt. Ihre erste Variable ist immer `self`. `self` stellt einfach immer das Objekt dar, welches initialisiert wird.

`beschleunigen` und `bremsen` sind 2 weitere Methoden. Auch sie enthalten immer als erstes Argument `self`. Von diesen Methoden könnte man so viele hinzufügen wie man möchte.

### Instanziierung eines Objekts
Wir können nun Objekte dieser Klasse instanziieren, zum Beispiel ein Auto:

```python
auto = Fahrzeug(4, "rot")
auto.bremsen()
```

Da wir bei den Parametern der `__init__`Funktion Standardwerte für die Anzahl der Räder (0) und die Farbe (schwarz) übergeben haben, können wir ein Fahrzeug ohne Parameter anzugeben erzeugen:

```python
schwarzes_auto_ohne_räder = Fahrzeug()
schwarzes_auto_ohne_räder.bremsen()
```

Wir können auch noch 2 weitere Fahrzeuge (S-Bahn und Motorrad) initialisieren:

```python
sbahn = Fahrzeug(64, "orange")
motorrad = Fahrzeug(2, "grau")
```

Jetzt können wir die Methoden der Objekte aufrufen, indem wir `objektname.methode()` eingeben:

```python
auto.beschleunigen()
auto.bremsen()
```

```python
schwarzes_auto_ohne_räder.beschleunigen()
schwarzes_auto_ohne_räder.bremsen()
```

```python
motorrad.beschleunigen()
motorrad.bremsen()
```

### Ändern der Attributwerte
In den Methoden können auch die Attributwerte des Objekts geändert werden. Damit kann das Objekt seinen internen Zustand ändern.

```python
# Die Klasse 'Fahrzeug' neu definiert und wird um ein Attribut 'Geschwindigkeit' erweitert
class Fahrzeug:
    def __init__(self, anzahl_räder=0, farbe="schwarz", geschwindigkeit=0):
        self.anzahl_räder = anzahl_räder
        self.farbe = farbe
        self.geschwindigkeit = geschwindigkeit

    def beschleunigen(self):
        self.geschwindigkeit += 1
        ausgabe = "Ich bin ein Fahrzeug mit Anzahl an Rädern: "
        ausgabe += str(self.anzahl_räder)
        ausgabe += " und Farbe: " + self.farbe
        ausgabe += " und ich beschleunige jetzt."
        ausgabe += "\n"
        ausgabe += "Meine neue Geschwindigkeit ist: "
        ausgabe += str(self.geschwindigkeit)
        print(ausgabe)

    def bremsen(self):
        self.geschwindigkeit -= 1
        ausgabe = "Ich bin ein Fahrzeug mit Anzahl an Rädern: "
        ausgabe += str(self.anzahl_räder)
        ausgabe += " und Farbe: " + self.farbe
        ausgabe += " und ich bremse jetzt."
        ausgabe += "\n"
        ausgabe += "Meine neue Geschwindigkeit ist: "
        ausgabe += str(self.geschwindigkeit) # ....
        print(ausgabe)

# teste deine Implementierung
motorrad = Fahrzeug(2, "grau")
motorrad.beschleunigen()
motorrad.beschleunigen()
motorrad.bremsen()
```

### Vererbung
Eine Klasse kann auch von einer anderen Klasse "erben". Das bedeutet, dass mehrere Fahrzeuge (z.B. Auto, S-Bahn, Motorrad) zwar manche Eigenschaften/Fähigkeiten gemeinsam haben (zB beschleunigen, bremsen), allerdings auch Eigenschaften/Fähigkeiten haben, die auf die anderen Fahrzeuge nicht zutrifft. Zum Beispiel kann nur ein Auto ein Schiebedach öffnen oder nur eine S-Bahn automatisch Türen öffnen.

Dafür gibt es die Möglichkeit, eine Klassenhierarchie aufzubauen. Es gibt eine **"Elternklasse"** (in diesem Fall das Fahrzeug), welche allgemeine Fähigkeiten (Methoden) und Eigenschaften (Attribute) mitbringt. In diesem Fall sind das die Fähigkeiten "beschleunigen" und "bremsen", und die Eigenschaften "Anzahl räder", "Farbe" und "Geschwindigkeit". Diese Fähigkeiten und Eigenschaften benötigen alle Fahrzeuge.

Neben der "Elternklasse", gibt es die **"Kindklassen"**, die die Fähigkeiten und Eigenschaften der Elternklasse übernehmen und erweitern. Das Auto erhält die Fähigkeit "Schiebedach öffnen". Die S-Bahn erhält die Fähigkeit, "automatisch Türen zu öffnen" und die Eigenschaft "Anzahl Waggons". Das Motorrad benötigt keine weiteren Fähigkeiten und erbt deshalb zwar die Attribute und Methoden der Elternklasse "Fahrzeug", implementiert jedoch keine neuen Methoden und instanziiert auch keine neuen Attribute. 

Man sagt: die Kindklassen **"erben"** von der Elternklasse und man spricht von **"Vererbung" von Eigenschaften und Fähigkeiten**:

```python
from IPython import display
display.Image("Vererbung.png")
```

Als Code würde das so aussehen:

```python
class Auto(Fahrzeug):
    def __init__(self, farbe):
        #Fahrzeug.__init__(self, anzahl_räder, farbe)
        super().__init__(4, farbe)
        self.schiebedach_offen = False
        
    def öffne_schiebedach(self):
        if not self.schiebedach_offen:
            print("Schiebedach wird geöffnet!")
            self.schiebedach_offen = True


class SBahn(Fahrzeug):
    def __init__(self, anzahl_räder, farbe, anzahl_waggons=6):
        #Fahrzeug.__init__(self, anzahl_räder, farbe)
        super().__init__(anzahl_räder, farbe)
        self.anzahl_waggons = anzahl_waggons

    def öffne_türen(self):
        print("Türen werden geöffnet!")


class Motorrad(Fahrzeug):
    def __init__(self, anzahl_räder, farbe):
        #Fahrzeug.__init__(self, anzahl_räder, farbe)
        super().__init__(anzahl_räder, farbe)
```

In der `__init__` Funktion wird nun die `__init__` Funktion der Elternklasse mit den entsprechenden Argumenten aufgerufen. Danach kann man auf die Attribute der Elternklasse zugreifen, als wären sie in der Kindklkasse definiert.

Nun werden die Objekte wieder instanziiert:

```python
auto1 = Auto("rot")
auto2 = Auto("grün")
sbahn1 = SBahn(64, "orange", 6)
sbahn2 = SBahn(64, "orange-rot", 4)
motorrad1 = Motorrad(2, "grau")
motorrad2 = Motorrad(2, "grau-schwarz")
```

Nun werden die Methoden der Objekte aufgerufen:

```python
#auto1.beschleunigen()
#motorrad2.bremsen()
sbahn1.öffne_türen()
auto1.öffne_schiebedach()
```

## X. Objektorientiertes Programmieren (OOP) vs. Funktionales Programmieren

Wenn man eine Aufgabe mit einem Programm lösen möchte, gibt es viele unterschiedliche Lösungsansätze um ans Ziel zu kommen.
Eine wichtige Unterscheidung in diesen Lösungen ist, ob sie **objektorientiert** oder **funktional** sind.
Im Folgenden sehen wir uns ein Beispiel für eine OOP bzw. funktionale Lösung für das selbe Problem an.

```python
# Objektorientiert
class Bankaccount:
    def __init__(self, besitzer, kontostand=0):
        self.besitzer = besitzer
        self.kontostand = kontostand

    def einzahlen(self, amount):
        self.kontostand += amount

    def auszahlen(self, betrag):
        if betrag > self.kontostand:
            print("Konto hat nicht genügend Guthaben!")
            return
        self.kontostand -= betrag

    def überweisen(self, anderes_konto, betrag):
        self.auszahlen(betrag)
        anderes_konto.einzahlen(betrag)

# Beispielfall
alice = Bankaccount("Alice", 100)
bob = Bankaccount("Bob", 50)

alice.überweisen(bob, 30)
print("Alice: " + str(alice.kontostand))
print("Bob: " + str(bob.kontostand))
```

```python
# Funktional
def erstelle_account(besitzer, kontostand=0):
    return {"besitzer": besitzer, "kontostand": kontostand}

def einzahlen(account, betrag):
    return {"besitzer": account["besitzer"], "kontostand": account["kontostand"] + betrag}

def auszahlen(account, betrag):
    if betrag > account["kontostand"]:
        print("Insufficient funds")
        return account
    return {"besitzer": account["besitzer"], "kontostand": account["kontostand"] - betrag}

def überweisen(von_account, zu_account, betrag):
    von_updated = auszahlen(von_account, betrag)
    zu_updated = einzahlen(zu_account, betrag)
    return von_updated, zu_updated

# Example usage
alice = erstelle_account("Alice", 100)
bob = erstelle_account("Bob", 50)

alice, bob = überweisen(alice, bob, 30)
print(alice)
print(bob)
```

**- Welche Variante ist leichter zu erweitern (also neue Attribute, Funktionen/Methoden, ...)?** \

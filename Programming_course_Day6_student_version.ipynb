**Tipp: In dieser Aufgabe werden *enums* sehr praktisch sein.** \
Enums (kurz für enumerations, also Aufzählungen) sind eine Menge von Namen,
die an Werte gebunden sind und im Code verwendet werden können. 
Enums sind in Python ein Spezialfall von Klassen, die keine Methoden haben. \
Anders als Variablen sind diese allerdings konstant. Hier ein Beispiel:
```
from enum import Enum

class Weekday(Enum):
    MONDAY = 1
    TUESDAY = 2
    WEDNESDAY = 3
    THURSDAY = 4
    FRIDAY = 5
    SATURDAY = 6
    SUNDAY = 7
    
print(Weekday.THURSDAY) # -> 4
```

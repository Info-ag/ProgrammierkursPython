## 2. Variablen und Strings
Eine **Variable** ist ein Name für einen Wert. Mit `=` speichert man einen Wert in einer Variablen:
```
name = "David"
```
Das `=` heißt hier nicht "ist gleich" wie in Mathe, sondern "*speichere den Wert rechts in der Variablen links*".
Danach kann man überall `name` schreiben und Python setzt den gespeicherten Wert ein.

Variablennamen schreibt man klein, ohne Leerzeichen (stattdessen `_`) und sie sollten beschreiben, was drin steht.

Strings kann man mit `+` aneinanderhängen. Achtung: `+` fügt **keine** Leerzeichen ein, die muss man selbst mit in die Strings schreiben.

````python
name = "David"
print(name)
````

````python
plz = "76131"
ort = "Karlsruhe"
print(ort)
ort_mit_plz = plz + " " + ort
print(ort_mit_plz)
````

````python
print("Mein Name ist " + name)
````

## 4. if-Bedingungen
Mit `if` kann ein Programm Entscheidungen treffen: Der eingerückte Code wird nur ausgeführt, wenn die Bedingung erfüllt ist.
```
if {Bedingung}:
    {Code, wenn die Bedingung erfüllt ist}
elif {andere Bedingung}:
    {Code, wenn die erste nicht, aber diese Bedingung erfüllt ist}
else:
    {Code, wenn keine der Bedingungen erfüllt ist}
```
- `elif` und `else` sind optional, `elif` darf beliebig oft vorkommen.
- **Wichtig:** Der Doppelpunkt `:` am Ende und die **Einrückung** (4 Leerzeichen bzw. `Tab`). Daran erkennt Python, welcher Code zum `if` gehört.
- Ob zwei Dinge gleich sind, prüft man mit `==` (zwei Gleichheitszeichen!). Ein einzelnes `=` speichert ja einen Wert in einer Variablen.

````python
antwort = input("Wie geht es dir?: ")
if antwort == "gut":
    print("Supi, mir geht es auch gut!")
elif antwort == "schlecht":
    print("Oh das tut mir leid")
else:
    print("Ich kann dich nicht verstehen.")
print("Das hier wird immer ausgegeben, weil es nicht eingerückt ist.")
````

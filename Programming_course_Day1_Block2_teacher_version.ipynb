## 5. Ganz- und (Fließ-)Kommazahlen
Neben Strings gibt es in Python auch Zahlen:
- **Integer** (`int`): Ganzzahlen wie `3`, `-17` oder `3000`
- **Float** (`float`): Kommazahlen wie `3.14` oder `-0.5`. **Achtung:** Python benutzt einen Punkt statt einem Komma!

Mit `type(...)` kann man herausfinden, welchen Datentyp ein Wert hat.

`input()` gibt immer einen **String** zurück, auch wenn man eine Zahl eintippt. Bevor man damit rechnen kann, muss man den String
mit `int(...)` bzw. `float(...)` in eine Zahl umwandeln. Umgekehrt macht `str(...)` aus einer Zahl einen String,
damit man sie mit `+` an einen anderen String hängen kann.

```python
stunden_programmiert = 3000     # Variable (alles klein geschrieben)
ANTWORT_AUF_ALLES = 42          # Konstante (gewöhnlich alles groß geschrieben)
PI = 3.14159
print(type(stunden_programmiert))
print(type(PI))
print(type("42"))
```

```python
# Weil input einen String zurückgibt, muss man das Ergebnis danach zuerst zu einem Integer konvertieren!
computer_alter = 17
alter = int(input("Wie alt bist du? "))

if computer_alter == alter:
    print("Lustig, ich auch")
elif computer_alter > alter:
    print("Dann bin ich " + str(computer_alter - alter) + " Jahre älter als du.")
else:
    print("Dann bist du", alter - computer_alter, "Jahre älter als ich")
```

## 6. Vergleichsoperationen und Bool'sche Operationen
Vergleiche ergeben entweder `True` (wahr) oder `False` (falsch). Diese beiden Werte nennt man **Boolean** (`bool`).

| Operator | Bedeutung |
|---|---|
| `==` | gleich |
| `!=` | ungleich |
| `<`, `>` | kleiner, größer |
| `<=`, `>=` | kleiner gleich, größer gleich |

Bedingungen kann man mit `and`, `or` und `not` kombinieren:
- `a and b` ist nur `True`, wenn **beide** `True` sind
- `a or b` ist `True`, wenn **mindestens eines** `True` ist
- `not a` ist das Gegenteil von `a`

```python
print("17 >= 17: ", 17 >= 17)
print("17 > 17: ", 17 > 17)
print("13 != 17: ", 13 != 17)
print("26 != 'Hallo': ", 26 != "Hallo") # Strings in Strings darstellen
print("'Hi' != 'Hi': ", "Hi" != "Hi")
```

## 7. Mathematische Operationen

| Operator | Bedeutung | Beispiel |
|---|---|---|
| `+`, `-`, `*`, `/` | Grundrechenarten | `7 / 2` → `3.5` |
| `**` | hoch | `2 ** 4` → `16` |
| `//` | Ganzzahldivision (immer abgerundet) | `7 // 2` → `3` |
| `%` | Modulo: der Rest beim Teilen | `7 % 2` → `1` |

Es gilt Punkt vor Strich, Klammern funktionieren wie in Mathe.
`a += 1` ist eine Abkürzung für `a = a + 1` (geht auch mit `-=`, `*=`, ...).

```python
a = 9
b = 12
print(a+b)
print(a-b)
print(a*b)
print(a/b) # Kommazahl
print(12*3+5)
print(12*(3+5))
print(2**4) # 'hoch'
```

```python
c = a/b
print(type(a))  # int
print(type(c))  # float

# Ganz- und Fließkommazahlen werden unterschiedlich gespeichert.
# In Python kann man trotzdem beide Datentypen problemlos miteinander verrechnen.
# Aber Achtung, Kommazahlen sind nicht immer ganz genau:
print(0.1 + 0.1 + 0.1)
print(0.1 + 0.1 + 0.1 == 0.3)
```

```python
print(10 / 3)  # Gleitkommadivision
print(10 // 3)  # Ganzzahldivision
print(10 % 3)   # Modulo, mathematischer Rest der Division

# Modulo Beispiel aus dem Alltag:
# 3 Personen teilen sich 10 Tortenstücke.
# Jede Person bekommt 3 Tortenstücke.
# Bleibt eine Torte als Rest übrig, das nicht fair aufgeteilt werden kann.
```

> **Mentoren:** Modulo ausführlich erklären, es kommt bei FizzBuzz, Quersumme und Caesar wieder.

### Aufgabe 7.2: Getränkeverkauf
Produkte: Wasser, Bier, Schnaps. \
Der Kunde soll eines der Produkte auswählen. Dann musst du ihn ggf. nach dem Alter fragen.
Du kannst ihm immer Wasser, Bier ab 16 Jahren und Schnaps ab 18 Jahren verkaufen.
Wenn du ihm etwas verkaufst oder nicht verkaufst, mache das mit einem `print` deutlich.

```python
produkt = input("Was möchtest du kaufen? ")

if produkt == "Wasser":
    print("Danke für deinen Einkauf!")
elif produkt == "Bier" or produkt == "Schnaps":
    alter = int(input("Wie alt bist du? "))
    if produkt == "Bier" and alter < 16:
        print("Du bist leider noch zu jung um Bier zu kaufen.")
    elif produkt == "Schnaps" and alter < 18:
        print("Du bist leider noch zu jung um Schnaps zu kaufen.")
    else:
        print("Danke für den Einkauf von " + produkt + "!")
else:
    print("Ich kenne das Produkt nicht.")
```

### Aufgabe 7.3: FizzBuzz
Frage nach einer Zahl. Wenn diese durch 3 **und** durch 5 teilbar ist, gib `FizzBuzz` aus,
sonst, wenn sie durch 3 teilbar ist `Fizz`, sonst wenn sie durch 5 teilbar ist `Buzz`,
und ansonsten einfach nur die Zahl.

```python
# Lösung 1
zahl = int(input("Gib hier die Zahl ein: "))
if zahl % 15 == 0:
    print("FizzBuzz")
elif zahl % 3 == 0:
    print("Fizz")
elif zahl % 5 == 0:
    print("Buzz")
else:
    print(zahl)

# Lösung 2
zahl = int(input("Gib hier die Zahl ein: "))
output = ""
if zahl % 3 == 0:
    output += "Fizz"
if zahl % 5 == 0:
    output += "Buzz"
if output == "":
    output = zahl
print(output)
```

> **Mentoren:** Bei FizzBuzz fragen, warum die Reihenfolge der Bedingungen wichtig ist (15 zuerst prüfen).

### Aufgabe 7.4: Schaltjahr (Zusatz)
Frage nach einer Jahreszahl und gib aus, ob es ein Schaltjahr ist. Ein Jahr ist ein Schaltjahr, wenn es durch 4 teilbar ist,
außer es ist durch 100 teilbar. Durch 400 teilbare Jahre sind aber trotzdem Schaltjahre. \
Beispiele: 2024 ✔, 1900 ✘, 2000 ✔, 2026 ✘

```python
jahr = int(input("Welches Jahr? "))
if jahr % 4 == 0 and (jahr % 100 != 0 or jahr % 400 == 0):
    print(jahr, "ist ein Schaltjahr")
else:
    print(jahr, "ist kein Schaltjahr")
```

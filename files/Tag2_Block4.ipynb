## Das Spiel
Wir programmieren **Hangman** (Galgenmännchen) **in Teams (z.B. zu zweit)**:
- Der Computer wählt zufällig ein geheimes Wort aus. Angezeigt wird für jeden Buchstaben ein Strich: `_ _ _ _ _ _`
- Man rät einen Buchstaben nach dem anderen. Kommt er im Wort vor, wird er an allen passenden Stellen aufgedeckt: `_ _ t _ _ n`
- Kommt er nicht vor, verliert man einen Versuch.
- Hat man alle Buchstaben gefunden, hat man gewonnen. Sind alle Versuche weg, hat man verloren.

So könnte das aussehen:
```
_ _ _ _ _ _
Noch 6 Versuche übrig.
Welchen Buchstaben rätst du? e
Leider falsch.

_ _ _ _ _ _
Noch 5 Versuche übrig.
Welchen Buchstaben rätst du? n
Richtig!

_ _ _ _ _ n
Noch 5 Versuche übrig.
...
```

**Ablauf:** Arbeitet zu in Teams. Baut das Spiel Schritt für Schritt aus kleinen Funktionen zusammen.

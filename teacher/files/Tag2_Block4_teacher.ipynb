## Das Spiel
Wir programmieren **Hangman** (Galgenmännchen) **in Teams (z.B. zu zweit)**:
- Der Computer wählt zufällig ein geheimes Wort aus. Angezeigt wird für jeden Buchstaben ein Strich: `_ _ _ _ _ _`
- Man rät einen Buchstaben nach dem anderen. Kommt er im Wort vor, wird er an allen passenden Stellen aufgedeckt: `_ _ t _ _ n`
- Kommt er nicht vor, verliert man einen Versuch.
- Hat man alle Buchstaben gefunden, hat man gewonnen. Sind alle Versuche weg, hat man verloren.

So könnte das aussehen:
```
_ _ _ _ _ _
Noch 6 Versuche übrig.
Welchen Buchstaben rätst du? e
Leider falsch.

_ _ _ _ _ _
Noch 5 Versuche übrig.
Welchen Buchstaben rätst du? n
Richtig!

_ _ _ _ _ n
Noch 5 Versuche übrig.
...
```

### Schritt 1: Ein Wort auswählen
Schreibe eine Funktion `waehle_wort()`, die ein zufälliges Wort aus der `WORTLISTE` zurückgibt. Du kannst die Liste gerne erweitern.

````python
import random

WORTLISTE = ["python", "programmieren", "schleife", "variable", "funktion", "liste", "computer", "tastatur", "heidelberg"]

def waehle_wort():
    return random.choice(WORTLISTE)

print(waehle_wort())
````

### Schritt 2: Das Wort anzeigen
Schreibe eine Funktion `zeige_wort(wort, geratene_buchstaben)`, die einen String zurückgibt, in dem alle bereits geratenen Buchstaben angezeigt werden
und alle anderen als `_`. Zwischen die Zeichen kommt jeweils ein Leerzeichen.

Beispiel: `zeige_wort("python", ["o", "p", "e"])` → `"p _ _ _ o _ "`

````python
def zeige_wort(wort, geratene_buchstaben):
    anzeige = ""
    for buchstabe in wort:
        if buchstabe in geratene_buchstaben:
            anzeige += buchstabe + " "
        else:
            anzeige += "_ "
    return anzeige

print(zeige_wort("python", ["o", "p", "e"]))
print(zeige_wort("python", []))
````

### Schritt 3: Ist das Wort erraten?
Schreibe eine Funktion `ist_erraten(wort, geratene_buchstaben)`, die `True` zurückgibt, wenn **alle** Buchstaben des Wortes schon geraten wurden, und sonst `False`.

````python
def ist_erraten(wort, geratene_buchstaben):
    for buchstabe in wort:
        if buchstabe not in geratene_buchstaben:
            return False
    return True

print(ist_erraten("liste", ["l", "i", "s", "t", "e", "x"]))  # True
print(ist_erraten("liste", ["l", "i"]))                      # False
````

### Schritt 4: Einen Buchstaben abfragen
Schreibe eine Funktion `frage_buchstabe(geratene_buchstaben)`, die so lange nach einem Buchstaben fragt, bis eine gültige Eingabe kommt, und diesen dann zurückgibt.
Gültig ist eine Eingabe, wenn sie
- genau ein Zeichen lang ist,
- ein Buchstabe von a bis z ist und
- noch nicht geraten wurde.

Bei einer ungültigen Eingabe soll eine passende Meldung ausgegeben werden. Großbuchstaben sollen auch funktionieren.

````python
def frage_buchstabe(geratene_buchstaben):
    while True:
        eingabe = input("Welchen Buchstaben rätst du? ").lower()
        if len(eingabe) != 1 or eingabe not in "abcdefghijklmnopqrstuvwxyz":
            print("Bitte gib genau einen Buchstaben ein.")
        elif eingabe in geratene_buchstaben:
            print("Den Buchstaben hast du schon geraten.")
        else:
            return eingabe   # return beendet die Funktion und damit auch die Schleife

print(frage_buchstabe(["a", "b"]))
````

### Schritt 5: Die Spielschleife
Jetzt setzen wir alles zusammen! Schreibe eine Funktion `hangman(versuche=6)`, die
1. ein Wort auswählt und eine leere Liste `geratene_buchstaben` anlegt,
2. so lange wiederholt, wie noch Versuche übrig sind **und** das Wort noch nicht erraten ist:
   - das Wort (mit `zeige_wort`) und die Anzahl der übrigen Versuche anzeigen
   - einen Buchstaben abfragen und zu `geratene_buchstaben` hinzufügen
   - ausgeben, ob er richtig oder falsch war, und bei falsch einen Versuch abziehen
3. am Ende ausgibt, ob man gewonnen oder verloren hat, und das Wort verrät.

````python
def hangman(versuche=6):
    wort = waehle_wort()
    geratene_buchstaben = []
    print("Willkommen bei Hangman!")

    while versuche > 0 and not ist_erraten(wort, geratene_buchstaben):
        print()
        print(zeige_wort(wort, geratene_buchstaben))
        print("Noch", versuche, "Versuche übrig.")
        buchstabe = frage_buchstabe(geratene_buchstaben)
        geratene_buchstaben.append(buchstabe)
        if buchstabe in wort:
            print("Richtig!")
        else:
            print("Leider falsch.")
            versuche -= 1

    if ist_erraten(wort, geratene_buchstaben):
        print("Gewonnen! Das Wort war:", wort)
    else:
        print("Verloren! Das Wort war:", wort)

hangman()
````

### Erweiterung A: Galgen zeichnen
Statt nur die Versuche herunterzuzählen, soll bei jedem Fehler ein Stück mehr vom Galgenmännchen gezeichnet werden.
Die Bilder haben wir schon für euch vorbereitet: `GALGEN[0]` ist das Bild bei 0 Fehlern, `GALGEN[6]` das bei 6 Fehlern.

Kopiert eure `hangman`-Funktion und baut die Bilder ein. *Tipp: Wie viele Fehler hat man gemacht, wenn man noch `versuche` von 6 Versuchen übrig hat?*

````python
GALGEN = [
"""
  +---+
  |   |
      |
      |
      |
      |
=========""",
"""
  +---+
  |   |
  O   |
      |
      |
      |
=========""",
"""
  +---+
  |   |
  O   |
  |   |
      |
      |
=========""",
"""
  +---+
  |   |
  O   |
 /|   |
      |
      |
=========""",
"""
  +---+
  |   |
  O   |
 /|\  |
      |
      |
=========""",
"""
  +---+
  |   |
  O   |
 /|\  |
 /    |
      |
=========""",
"""
  +---+
  |   |
  O   |
 /|\  |
 / \  |
      |
=========""",
]

def hangman_mit_galgen():
    max_versuche = len(GALGEN) - 1
    versuche = max_versuche
    wort = waehle_wort()
    geratene_buchstaben = []
    print("Willkommen bei Hangman!")

    while versuche > 0 and not ist_erraten(wort, geratene_buchstaben):
        print(GALGEN[max_versuche - versuche])
        print(zeige_wort(wort, geratene_buchstaben))
        buchstabe = frage_buchstabe(geratene_buchstaben)
        geratene_buchstaben.append(buchstabe)
        if buchstabe in wort:
            print("Richtig!")
        else:
            print("Leider falsch.")
            versuche -= 1

    print(GALGEN[max_versuche - versuche])
    if ist_erraten(wort, geratene_buchstaben):
        print("Gewonnen! Das Wort war:", wort)
    else:
        print("Verloren! Das Wort war:", wort)

hangman_mit_galgen()
````

### Erweiterung B: Falsche Buchstaben anzeigen
Zeige in jeder Runde zusätzlich an, welche falschen Buchstaben schon geraten wurden, z.B. `Falsch: e, a, r`.
*Tipp: `", ".join(liste)` verbindet alle Strings einer Liste mit `", "` dazwischen.*

````python
def falsche_buchstaben(wort, geratene_buchstaben):
    falsche = []
    for buchstabe in geratene_buchstaben:
        if buchstabe not in wort:
            falsche.append(buchstabe)
    return falsche

print("Falsch:", ", ".join(falsche_buchstaben("python", ["e", "p", "a", "r"])))
# In der Spielschleife nach zeige_wort(...) einfach diese Zeile einfügen.
````

### Erweiterung C: Hangman als Klasse
Schreibt das Spiel als Klasse `Hangman` mit den Attributen `wort`, `geratene_buchstaben` und `versuche` und z.B. den Methoden
`anzeige()`, `raten(buchstabe)`, `ist_erraten()`, `ist_verloren()` und `spielen()`.
Ihr könnt dabei eure Funktionen von oben wiederverwenden.

````python
class Hangman:
    def __init__(self, wort, versuche=6):
        self.wort = wort
        self.geratene_buchstaben = []
        self.versuche = versuche

    def anzeige(self):
        return zeige_wort(self.wort, self.geratene_buchstaben)

    def raten(self, buchstabe):
        self.geratene_buchstaben.append(buchstabe)
        if buchstabe in self.wort:
            print("Richtig!")
        else:
            print("Leider falsch.")
            self.versuche -= 1

    def ist_erraten(self):
        return ist_erraten(self.wort, self.geratene_buchstaben)

    def ist_verloren(self):
        return self.versuche == 0

    def spielen(self):
        while not self.ist_erraten() and not self.ist_verloren():
            print()
            print(self.anzeige())
            print("Noch", self.versuche, "Versuche übrig.")
            self.raten(frage_buchstabe(self.geratene_buchstaben))
        if self.ist_erraten():
            print("Gewonnen! Das Wort war:", self.wort)
        else:
            print("Verloren! Das Wort war:", self.wort)

spiel = Hangman(waehle_wort())
spiel.spielen()
````

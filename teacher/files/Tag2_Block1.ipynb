## 13. import und random
Python bringt viele fertige Werkzeuge mit, die in **Modulen** stecken. Mit `import modulname` lädt man ein Modul,
danach kann man seine Funktionen mit `modulname.funktion(...)` benutzen.

Das Modul `random` erzeugt Zufallszahlen:
- `random.randint(a, b)`: zufällige ganze Zahl von `a` bis `b` (beide inklusive)
- `random.choice(liste)`: zufälliges Element aus einer Liste

````python
import random

print(random.randint(1, 6))
print(random.choice(["Schere", "Stein", "Papier"]))
````

````python
r = random.randint(1, 10)
if int(input("Rate meine Zahl zwischen 1 und 10: ")) == r:
    print("Richtig, super geraten!")
else:
    print("Das ist leider nicht meine Zahl, es war die", r)
````

## 14. Funktionen
Funktionen haben wir schon die ganze Zeit benutzt: `print()`, `input()`, `len()`, `random.randint()`, ...
Mit `def` kann man auch eigene Funktionen schreiben. Eine Funktion ist ein Stück Code mit einem Namen,
das man so oft man will **aufrufen** kann, ohne den Code jedes Mal neu zu schreiben.
```
def funktionsname(parameter1, parameter2):
    {Code}
    return {Ergebnis}
```
- **Parameter** sind Variablen, die beim Aufruf einen Wert bekommen: `funktionsname(3, "Hallo")`.
- Mit `return` gibt die Funktion ein Ergebnis zurück, das man z.B. in einer Variablen speichern kann.
  Nach `return` ist die Funktion sofort zu Ende.
- Der Code in der Funktion wird erst ausgeführt, wenn die Funktion aufgerufen wird, nicht schon beim `def`!

````python
def sag_hallo():
    print("Hallo!")

sag_hallo()
sag_hallo()
````

````python
# Funktionen verkürzen, verschönern und vereinfachen Code
def vorstellen():
    print("Hallo, ich bin David.")
    print("Ich studiere in Karlsruhe.")
    print("Ich mag Informatik.")

vorstellen()
vorstellen() # wenn man sich mehrfach vorstellen muss, muss man nur die Funktion mehrfach aufrufen
````

````python
# Parameter
def vorstellen2(name):
    print("Hallo, ich bin " + name + ".")

vorstellen2("David")
vorstellen2("Marvin")
````

````python
# return ist sehr nützlich, wenn man das Ergebnis weiterverwenden will
def f(x):
    return 2*x**2 + 3*x - 5

y = f(1)
print(y)
print(f(3) + f(4))
````

````python
# Es gehen auch mehrere Parameter
def g(x, y):
    return (x + y)**2

print(g(2, 3))
print(g(7, 2))
````

**print vs. return:** `print` zeigt etwas auf dem Bildschirm an, damit kann das Programm danach aber nicht weiterrechnen.
`return` gibt den Wert an die Stelle zurück, an der die Funktion aufgerufen wurde.

````python
def doppelt_print(x):
    print(2 * x)

def doppelt_return(x):
    return 2 * x

a = doppelt_print(5)   # gibt 10 aus, aber a ist None (nichts)
b = doppelt_return(5)  # gibt nichts aus, aber b ist 10
print(a, b)
````

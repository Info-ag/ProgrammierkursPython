```python
class Fahrzeug:
    def __init__(self, anzahl_räder=0, farbe="schwarz"):
        self.anzahl_räder = anzahl_räder
        self.farbe = farbe
        self.geschwindigkeit = 0

    def beschleunigen(self):
        self.geschwindigkeit += 10
        print("Ich beschleunige, meine Geschwindigkeit ist jetzt", self.geschwindigkeit)

    def bremsen(self):
        self.geschwindigkeit -= 10
        print("Ich bremse, meine Geschwindigkeit ist jetzt", self.geschwindigkeit)


class Tier:
    def __init__(self, name, alter):
        self.name = name
        self.alter = alter
        self.gelaufene_schritte = 0

    def vorstellen(self):
        print("Ich heiße " + self.name + " und bin " + str(self.alter) + " Jahre alt.")

    def laufen(self, anzahl_schritte):
        print("Ich laufe " + str(anzahl_schritte) + " Schritte.")
        self.gelaufene_schritte += anzahl_schritte
```

### Vererbung
Eine Klasse kann auch von einer anderen Klasse "erben". Mehrere Fahrzeuge (z.B. Auto, S-Bahn, Motorrad) haben manche Eigenschaften/Fähigkeiten gemeinsam (z.B. beschleunigen, bremsen),
aber auch Eigenschaften/Fähigkeiten, die auf die anderen Fahrzeuge nicht zutreffen: Nur ein Auto kann ein Schiebedach öffnen und nur eine S-Bahn hat Waggons.

Die **Elternklasse** (`Fahrzeug`) bringt die allgemeinen Attribute und Methoden mit. Die **Kindklassen** übernehmen ("erben") diese und können sie erweitern
oder Methoden mit gleichem Namen neu schreiben (überschreiben).

![Vererbung](https://raw.githubusercontent.com/Info-ag/ProgrammierkursPython/2days/images/Vererbung.png)

```python
class Auto(Fahrzeug):          # Auto erbt von Fahrzeug
    def __init__(self, farbe):
        super().__init__(4, farbe)   # ruft __init__ der Elternklasse auf
        self.schiebedach_offen = False

    def öffne_schiebedach(self):
        if not self.schiebedach_offen:
            print("Schiebedach wird geöffnet!")
            self.schiebedach_offen = True


class SBahn(Fahrzeug):
    def __init__(self, farbe, anzahl_waggons=6):
        super().__init__(8 * anzahl_waggons, farbe)
        self.anzahl_waggons = anzahl_waggons

    def öffne_türen(self):
        print("Türen werden geöffnet!")


auto1 = Auto("rot")
sbahn1 = SBahn("orange", 4)
auto1.beschleunigen()    # geerbt von Fahrzeug
auto1.öffne_schiebedach()
sbahn1.öffne_türen()
print(sbahn1.anzahl_räder)
```

### Aufgabe 16.2
Schreibe zwei Kindklassen `Hund` und `Vogel`, die von `Tier` erben.
- `Hund` bekommt das zusätzliche Attribut `rasse`. `vorstellen()` soll jetzt ausgeben: "Ich heiße [Name], bin ein [Rasse] und bin [Alter] Jahre alt."
- `Vogel` bekommt das zusätzliche Attribut `farbe` und die zusätzliche Methode `fliegen(meter)`, die "Ich fliege [Meter] Meter weit." ausgibt.

```python
class Hund(Tier):
    def __init__(self, name, alter, rasse):
        super().__init__(name, alter)
        self.rasse = rasse

    def vorstellen(self):
        print("Ich heiße " + self.name + ", bin ein " + self.rasse + " und bin " + str(self.alter) + " Jahre alt.")


class Vogel(Tier):
    def __init__(self, name, alter, farbe):
        super().__init__(name, alter)
        self.farbe = farbe

    def fliegen(self, meter):
        print("Ich fliege " + str(meter) + " Meter weit.")

hund = Hund("Maya", 13, "Cocker-Spaniel")
hund.vorstellen()
hund.laufen(10)
vogel = Vogel("Rubi", 10, "rot")
vogel.vorstellen()
vogel.fliegen(100)
```

## 17. Objektorientiert vs. funktional
Wenn man eine Aufgabe mit einem Programm lösen möchte, gibt es viele unterschiedliche Lösungsansätze.
Eine wichtige Unterscheidung ist, ob sie **objektorientiert** (mit Klassen) oder **funktional** (mit Funktionen und z.B. Dictionaries) sind.
Hier ein Beispiel für beide Varianten für dasselbe Problem:

```python
# Objektorientiert
class Bankkonto:
    def __init__(self, besitzer, kontostand=0):
        self.besitzer = besitzer
        self.kontostand = kontostand

    def einzahlen(self, betrag):
        self.kontostand += betrag

    def auszahlen(self, betrag):
        if betrag > self.kontostand:
            print("Konto hat nicht genügend Guthaben!")
            return
        self.kontostand -= betrag

    def überweisen(self, anderes_konto, betrag):
        self.auszahlen(betrag)
        anderes_konto.einzahlen(betrag)

alice = Bankkonto("Alice", 100)
bob = Bankkonto("Bob", 50)
alice.überweisen(bob, 30)
print("Alice:", alice.kontostand)
print("Bob:", bob.kontostand)
```

```python
# Funktional
def erstelle_konto(besitzer, kontostand=0):
    return {"besitzer": besitzer, "kontostand": kontostand}

def einzahlen(konto, betrag):
    return {"besitzer": konto["besitzer"], "kontostand": konto["kontostand"] + betrag}

def auszahlen(konto, betrag):
    if betrag > konto["kontostand"]:
        print("Konto hat nicht genügend Guthaben!")
        return konto
    return {"besitzer": konto["besitzer"], "kontostand": konto["kontostand"] - betrag}

def überweisen(von_konto, zu_konto, betrag):
    return auszahlen(von_konto, betrag), einzahlen(zu_konto, betrag)

alice = erstelle_konto("Alice", 100)
bob = erstelle_konto("Bob", 50)
alice, bob = überweisen(alice, bob, 30)
print(alice)
print(bob)
```

Was haltet ihr von den beiden Implementierungen? \
**- Welche Variante ist einfacher zu verstehen?** \
**- Welche Variante ist leichter zu erweitern (also neue Attribute, Funktionen/Methoden, ...)?** \
**- Welche Variante würdet ihr in einer Bank-App bevorzugen?**

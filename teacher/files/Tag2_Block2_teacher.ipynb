## 15. Funktionen kombinieren
Richtig nützlich werden Funktionen, wenn man größere Probleme in kleine Teile zerlegt:
Jede Funktion löst eine kleine Aufgabe und Funktionen können andere Funktionen aufrufen.
Genau so werden wir heute Nachmittag das Projekt bauen.

Ein paar Dinge, die man dabei wissen sollte:
- **Standardwerte:** Parameter können einen Standardwert bekommen (`def f(x, y=2):`). Dann darf man sie beim Aufruf weglassen.
- **Geltungsbereich:** Variablen, die *in* einer Funktion angelegt werden (inklusive der Parameter), gibt es nur in dieser Funktion.
  Will man ein Ergebnis außerhalb benutzen, muss man es mit `return` zurückgeben.
- Funktionen können Listen und Strings als Parameter bekommen und zurückgeben.

```python
def reverse(string):
    rev_string = ""
    for i in range(1, len(string)+1):
        rev_string += string[-i]
    return rev_string

def ist_palindrom(wort):
    return wort == reverse(wort)   # ruft die Funktion reverse auf

print(reverse("Hallo!"))
print(ist_palindrom("racecar"))
print(ist_palindrom("python"))
```

```python
# Standardwerte
def potenz(basis, exponent=2):
    return basis ** exponent

print(potenz(5))     # exponent ist 2
print(potenz(5, 3))  # exponent ist 3
```

```python
# Geltungsbereich
def verdoppeln(zahl):
    ergebnis = zahl * 2
    return ergebnis

x = verdoppeln(21)
print(x)
# print(ergebnis)  # Fehler! ergebnis gibt es nur innerhalb der Funktion
```

### Aufgabe 15.1
Schreibe eine Funktion `min_of_list(liste)`, die das kleinste Element einer Liste von Zahlen zurückgibt (ohne die eingebaute Funktion `min` zu benutzen).

```python
def min_of_list(liste):
    kleinstes = liste[0]
    for zahl in liste:
        if zahl < kleinstes:
            kleinstes = zahl
    return kleinstes

print(min_of_list([5, 776, 2, 43, -6, 13]))
```

### Aufgabe 15.2
Schreibe eine Funktion `zaehle_buchstabe(text, buchstabe)`, die zurückgibt, wie oft `buchstabe` in `text` vorkommt.
Schreibe dann eine Funktion `haeufigster_buchstabe(text)`, die `zaehle_buchstabe` benutzt, um den häufigsten Buchstaben zu finden.

```python
def zaehle_buchstabe(text, buchstabe):
    anzahl = 0
    for b in text:
        if b == buchstabe:
            anzahl += 1
    return anzahl

def haeufigster_buchstabe(text):
    bester = text[0]
    for b in text:
        if zaehle_buchstabe(text, b) > zaehle_buchstabe(text, bester):
            bester = b
    return bester

print(zaehle_buchstabe("banane", "n"))
print(haeufigster_buchstabe("mississippi"))
```

### Aufgabe 15.3: rotate_string
Schreibe eine Funktion `rotate_string(s, k)`, die einen String `s` um `k` Stellen rotiert.
Bei einer Rotation um 1 wird das letzte Zeichen an den Anfang gesetzt, bei einer Rotation um `k` wird das `k` mal gemacht.
`rotate_string("Hallo", 2)` → `"loHal"`

*Tipp: Du kannst, musst aber nicht, Slices verwenden.*
- Einfache Version: `k <= len(s)`
- Schwerere Version: `k` darf größer als `len(s)` sein (*Tipp: `%`*)

```python
# Einfache Version
def rotate_string(s, k):
    return s[-k:] + s[:-k]

# Mit Schleife (funktioniert auch für k > len(s), wird für riesige k aber langsam)
def rotate_string(s, k):
    for i in range(k):
        s = s[-1] + s[:-1]
    return s

# Schwerere Version
def rotate_string(s, k):
    k = k % len(s)
    return s[-k:] + s[:-k]

print(rotate_string("Hallo", 2))
print(rotate_string("Hallo", 12))
```

### Aufgabe 15.4: Caesar mit Funktionen (Zusatz)
Schreibe die Caesar-Verschlüsselung von gestern als zwei Funktionen `caesar_enc(nachricht, k)` und `caesar_dec(nachricht, k)`.
*Tipp: Die Entschlüsselung ist eine Verschlüsselung mit `-k`.*

```python
alph = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"

def caesar_enc(nachricht, k):
    ergebnis = ""
    for buchstabe in nachricht:
        ergebnis += alph[(alph.index(buchstabe) + k) % 26]
    return ergebnis

def caesar_dec(nachricht, k):
    return caesar_enc(nachricht, -k)

enc = caesar_enc("DIEGERMANENGREIFENAN", 2)
print(enc)
print(caesar_dec(enc, 2))
```

## 16. Klassen und Objekte
### Objekte
Ein **Objekt** hat **Eigenschaften** (Dinge, die es besitzt) und **Fähigkeiten** (Dinge, die es kann).
- Die Eigenschaften werden auch **Attribute** genannt und sie werden meistens mit Substantiven bezeichnet.
- Die Fähigkeiten eines Objekts sind seine **Methoden** (einfach Funktionen, die zu einem Objekt gehören). Sie werden meistens mit Verben benannt.

Anhand eines Fahrzeugs kann man sich das so vorstellen:

![Klassen Übersicht](https://raw.githubusercontent.com/Info-ag/ProgrammierkursPython/2days/images/Klassen_%C3%9Cbersicht.png)

### Klassen
Eine Klasse ist eine Art Bauplan, mit dem Objekte erstellt werden können. Im Programmcode definiert man eine Klasse und dann lässt man Objekte dieser Klasse zusammenbauen.
Das Zusammenbauen der Objekte nennt man auch **"Instanziieren"**. Je nachdem, welche Werte man beim Instanziieren mitgibt, können verschiedene Objekte erstellt werden.

Alle Objekte einer Klasse haben die gleichen Attribute und die gleichen Methoden. **Die Werte der Attribute können sich aber zwischen den Objekten unterscheiden.**
Beispielsweise haben ein Auto und ein Motorrad beide Räder, aber unterschiedlich viele. In einem Objekt sind also die Daten (Attribute) und Fähigkeiten (Methoden) zusammengefasst.

### Definition einer Klasse

```python
class Fahrzeug:
    """Ein Fahrzeug hat eine Anzahl an Rädern und eine Farbe.
    Außerdem kann es beschleunigen und bremsen."""
    def __init__(self, anzahl_räder=0, farbe="schwarz"):
        self.anzahl_räder = anzahl_räder
        self.farbe = farbe

    def beschleunigen(self):
        print("Ich bin ein " + self.farbe + "es Fahrzeug mit " + str(self.anzahl_räder) + " Rädern und ich beschleunige jetzt.")

    def bremsen(self):
        print("Ich bin ein " + self.farbe + "es Fahrzeug mit " + str(self.anzahl_räder) + " Rädern und ich bremse jetzt.")
```

Unsere Klasse hat den Namen **Fahrzeug** und beinhaltet die 3 Methoden `__init__`, `beschleunigen` und `bremsen`.

- Die Methode `__init__` wird beim Instanziieren des Objekts aufgerufen, sie "initialisiert" also das Objekt. Hier werden die Attribute angelegt.
- Der erste Parameter jeder Methode ist immer `self`. `self` ist das Objekt selbst. Mit `self.attribut` greift man auf die Attribute des Objekts zu.
- `beschleunigen` und `bremsen` sind 2 weitere Methoden. Davon kann man so viele hinzufügen wie man möchte.
- Klassennamen schreibt man mit großem Anfangsbuchstaben.

### Instanziierung eines Objekts
Wir können nun Objekte dieser Klasse instanziieren, indem wir die Klasse wie eine Funktion aufrufen.
Die Methoden eines Objekts ruft man mit `objektname.methode()` auf, `self` gibt man dabei nicht mit an.

```python
auto = Fahrzeug(4, "rot")
motorrad = Fahrzeug(2, "grau")
auto.beschleunigen()
motorrad.bremsen()
print(auto.farbe)
```

```python
# Da __init__ Standardwerte hat, geht es auch ohne Parameter:
schwarzes_fahrzeug_ohne_räder = Fahrzeug()
schwarzes_fahrzeug_ohne_räder.bremsen()
```

### Ändern der Attributwerte
In den Methoden können auch die Attributwerte des Objekts geändert werden. Damit kann das Objekt seinen internen Zustand ändern.

```python
# Die Klasse 'Fahrzeug' wird neu definiert und um ein Attribut 'geschwindigkeit' erweitert
class Fahrzeug:
    def __init__(self, anzahl_räder=0, farbe="schwarz"):
        self.anzahl_räder = anzahl_räder
        self.farbe = farbe
        self.geschwindigkeit = 0

    def beschleunigen(self):
        self.geschwindigkeit += 10
        print("Ich beschleunige, meine Geschwindigkeit ist jetzt", self.geschwindigkeit)

    def bremsen(self):
        self.geschwindigkeit -= 10
        print("Ich bremse, meine Geschwindigkeit ist jetzt", self.geschwindigkeit)

motorrad = Fahrzeug(2, "grau")
auto = Fahrzeug(4, "rot")
motorrad.beschleunigen()
motorrad.beschleunigen()
motorrad.bremsen()
auto.beschleunigen()  # jedes Objekt hat seine eigene Geschwindigkeit
```

### Aufgabe 16.1
Schreibe eine Klasse `Tier` mit den Attributen `name`, `alter` und `gelaufene_schritte` (am Anfang 0) und folgenden Methoden:
- `vorstellen()` soll ausgeben: "Ich heiße [Name] und bin [Alter] Jahre alt."
- `laufen(anzahl_schritte)` soll ausgeben: "Ich laufe [Anzahl Schritte] Schritte." \
    Außerdem soll `gelaufene_schritte` um `anzahl_schritte` erhöht werden.

Erstelle zwei Tiere und teste alle Methoden.

```python
class Tier:
    def __init__(self, name, alter):
        self.name = name
        self.alter = alter
        self.gelaufene_schritte = 0

    def vorstellen(self):
        print("Ich heiße " + self.name + " und bin " + str(self.alter) + " Jahre alt.")

    def laufen(self, anzahl_schritte):
        print("Ich laufe " + str(anzahl_schritte) + " Schritte.")
        self.gelaufene_schritte += anzahl_schritte

tier1 = Tier("Maja", 3)
tier2 = Tier("Rex", 7)
tier1.vorstellen()
tier2.vorstellen()
tier1.laufen(5)
tier1.laufen(2)
print(tier1.gelaufene_schritte)
```

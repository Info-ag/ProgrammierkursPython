## 11. Slices und Löschen
*Slices* sind eine fortgeschrittene Variante von Indexen.
Mit einem Index wählt man *ein* Element aus, mit einem Slice wählt man eine *Teilliste* aus.

Slices schreibt man wie `range` mit Start, Ende und Schrittweite, nur mit Doppelpunkten: `liste[start:ende:schritt]`.
- Das Ende ist (wie bei `range`) nicht mit eingeschlossen.
- Lässt man den Start weg, wird vom Anfang an genommen, lässt man das Ende weg, bis zum Schluss.
- Slices funktionieren auch mit Strings.

````python
list5 = [10, 20, 30, 40, 50, 60]
print(list5[:3])   # Indexe 0,1,2
print(list5[3:])   # Indexe 3,4,5
print(list5[2:4])  # Indexe 2,3
print(list5[::2])  # jedes zweite Element
print(list5[::-1]) # rückwärts!
````

````python
# Umgekehrt kommt man mit index() vom Element auf den Index.
# Wenn das Element mehrmals vorkommt, wird der niedrigste Index zurückgegeben.
print(list5.index(30))
alph = "abcdefghijklmnopqrstuvwxyz"
print(alph.index("l"))
````

````python
# Wie man Elemente aus Listen löscht:
list6 = [3, 6, 8, 2, 13, 67, 24]
list6.pop(-2)     # Löschen mit Index
print(list6)
list6.remove(8)   # Löschen mit Element
print(list6)
del list6[1:3]    # Löschen mit Slice
print(list6)
````

## 12. Dictionaries
Ein **Dictionary** (`dict`) speichert Paare aus **Schlüssel** (*key*) und **Wert** (*value*), wie ein Wörterbuch.
Statt über einen Index (0, 1, 2, ...) greift man über den Schlüssel auf den Wert zu.
```
preise = {"Apfel": 1, "Banane": 2}
preise["Apfel"]        # -> 1
preise["Kiwi"] = 0.5   # neues Paar hinzufügen (oder Wert ändern)
```
- `dict.keys()`, `dict.values()` und `dict.items()` geben die Schlüssel, die Werte bzw. die Paare zurück.
- Mit `in` prüft man, ob ein **Schlüssel** im Dictionary vorkommt.
- Jeder Schlüssel kommt höchstens einmal vor.

````python
prices = {"Apple": 1, "Banana": 2, "Strawberries": 3, "Potato": 0.5}
buying = {"Apple": 3, "Banana": 5, "Potato": 10}

print(prices["Apple"])
print(buying.keys())
print(buying.values())
print(buying.items())
````

````python
prices["Carrot"] = 1.5
print("Carrot" in prices)
for key, value in prices.items():
    print("The product", key, "costs", value, "€.")
````
